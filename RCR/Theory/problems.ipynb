```python
import clingo

def resolver_asp(titulo, codigo_asp, opciones=["0"]):
    """
    Función de soporte para instanciar, ejecutar y procesar modelos estables en Clingo.
    - titulos: Cadena de texto para etiquetar y organizar visualmente los resultados
    - codigo_asp: Cadena de texto con el código en ASP escrito en la sintaxis formal de Clingo/Gringo
    - opciones: Lista de parámetros de Clingo (ej. ["0"] indica a Clingo que calcule y devuelva todos los modelos)
    """
    print("=" * 74)
    print(f"  {titulo.upper()}")
    print("=" * 74)

    errores = []

    # Captura los mensajes generados por Clingo
    def logger(codigo, mensaje):
        errores.append(mensaje)

    ctl = clingo.Control(opciones, logger=logger)
    try:
        ctl.add("base", [], codigo_asp)
        ctl.ground([("base", [])])
    except RuntimeError:
        print("\n❌ ERROR EN EL PROGRAMA ASP\n")
        for error in errores:
            print(error)
        return

    modelos = []
    with ctl.solve(yield_=True) as handle:
        for modelo in handle:
            modelos.append(modelo.symbols(shown=True))

    if len(modelos) == 0:
        print("  [ UNSATISFIABLE ] No existe ninguna solución que cumpla las restricciones.\n")
        return

    print(f"\nNúmero de modelos: {len(modelos)}\n")
    for i, modelo in enumerate(modelos, 1):
        print(f"Modelo {i}:")
        print(modelo)
```

**Ejercicio 1.** 
Animales y alimentación Representa en ASP el siguiente conocimiento: • Toby y Max son perros. • Luna es un gato. • Todo perro es un animal.Todo gato es un animal.Toby come pienso y Luna come pescado. El programa debe permitir obtener qué individuos son animales y qué alimentos consume cada uno. Muestra únicamente los predicados animal/1 y come/2. Número de modelos estables esperado: 1. 

```python
ej = ''' 
% Hechos
perro(toby; max).
gato(luna).

% Reglas
animal(X) :- perro(X).
animal(X) :- gato(X).

come(toby, pienso).
come(luna, pescado).

% Directivas de salida
#show animal/1.
#show come/2.
'''

resolver_asp("Animales y alimentacion", ej)
```

**Ejercicio 2.** 
Acceso a una biblioteca Una biblioteca registra los siguientes datos: Ana, Luis y Marta son usuarios. Ana y Marta tienen carnet. Un usuario puede acceder a la biblioteca si tiene carnet. Modela los hechos y la regla que permite determinar qué usuarios pueden acceder. Añade una regla para indicar que un usuario necesita renovar el carnet si es usuario pero no consta que tenga carnet. Muestra puede_acceder/1 y renovar/1. Número de modelos estables esperado: 1. 

```python
ej = ''' 
% Hechos
usario(marta; ana; luis).
carnet(ana; marta).

% Reglas
puede_acceder(X) :- carnet(X), usario(X).
renovar(X) :- usario(X), not carnet(X).

% Salida
#show puede_acceder/1.
#show renovar/1.

'''

resolver_asp("Acceso a una biblioteca", ej)
```
